```python
from pathlib import Path

db_path = Path("../rag/data/chat_history.db")

print("DB exists:", db_path.exists())
print("DB path:", db_path.resolve())
print("DB size:", db_path.stat().st_size if db_path.exists() else "N/A")
```

```python
import sqlite3

conn = sqlite3.connect(db_path)
tables = conn.execute("""
    SELECT name
    FROM sqlite_master
    WHERE type = 'table'
    ORDER BY name;
""").fetchall()

tables
```

```python
schema = conn.execute("""
    SELECT sql
    FROM sqlite_master
    WHERE type = 'table'
    AND name = 'sessions';                     
""").fetchall()[0]

print(schema)
```

```python
import pandas as pd

df_sessions = pd.read_sql_query("""
    SELECT *
    FROM sessions
    ORDER BY created_at DESC
    LIMIT 10
""", conn)

display(df_sessions)
```
